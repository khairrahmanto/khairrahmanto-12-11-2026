```python
import pandas as pd

print("Pandas berhasil diimport!")
```

```python
df = pd.read_csv('/content/sales_detail.csv')

print("Dataset berhasil dibaca!")
```

```python
print("Jumlah data kosong setiap kolom:")
print(df.isnull().sum())
```

```python
print("Jumlah data duplikat:")
print(df.duplicated().sum())
```

```python
df['Date'] = pd.to_datetime(df['Date'], format='%d-%b-%y')

print("Tipe data Date setelah diperbaiki:")
print(df['Date'].dtype)
```

```python
df.info()
```

```python
# Statistik dasar penjualan

print("TOTAL PENJUALAN:")
print(df['Value'].sum())

print("\nRATA-RATA TRANSAKSI:")
print(df['Value'].mean())

print("\nMEDIAN PENJUALAN:")
print(df['Value'].median())

print("\nSTANDAR DEVIASI:")
print(df['Value'].std())

print("\nPENJUALAN TERENDAH:")
print(df['Value'].min())

print("\nPENJUALAN TERTINGGI:")
print(df['Value'].max())
```

```python
# Total penjualan berdasarkan kategori

penjualan_kategori = df.groupby('Category')['Value'].sum()

penjualan_kategori = penjualan_kategori.sort_values(
    ascending=False
)

print(penjualan_kategori)
```

```python
# Jumlah barang terjual berdasarkan produk

produk_terlaris = df.groupby('Product Name')['Qty'].sum()

produk_terlaris = produk_terlaris.sort_values(
    ascending=False
)

print("5 PRODUK TERLARIS:")
print(produk_terlaris.head(5))
```

```python
df[['Qty', 'Price', 'Value']].describe()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

total_kategori = df.groupby('Category')['Value'].sum()
total_kategori = total_kategori.sort_values(ascending=False)

plt.figure(figsize=(12, 6))

sns.barplot(
    x=total_kategori.values,
    y=total_kategori.index,
    color='skyblue'
)

plt.title('Total Penjualan per Kategori Produk')
plt.xlabel('Total Penjualan (Rp)')
plt.ylabel('Kategori Produk')

plt.ticklabel_format(style='plain', axis='x')
plt.tight_layout()
plt.show()
```

```python
df['Month'] = df['Date'].dt.to_period('M')

penjualan_bulanan = df.groupby('Month')['Value'].sum()

plt.figure(figsize=(12, 6))

plt.plot(
    penjualan_bulanan.index.astype(str),
    penjualan_bulanan.values,
    marker='o'
)

plt.title('Tren Penjualan per Bulan')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan (Rp)')
plt.xticks(rotation=45)
plt.ticklabel_format(style='plain', axis='y')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.scatterplot(
    data=df,
    x='Qty',
    y='Value',
    alpha=0.5
)

plt.title('Hubungan Jumlah Produk dan Nilai Penjualan')
plt.xlabel('Jumlah Produk Terjual')
plt.ylabel('Nilai Penjualan (Rp)')

plt.ticklabel_format(style='plain', axis='y')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df['Price'], bins=30, edgecolor='black')

plt.title('Distribusi Harga Produk')
plt.xlabel('Harga Produk (Rp)')
plt.ylabel('Frekuensi')

plt.ticklabel_format(style='plain', axis='x')
plt.tight_layout()
plt.show()
```

```python

plt.figure(figsize=(10, 6))

sns.boxplot(x=df['Value'], color='skyblue')

plt.title('Boxplot Nilai Penjualan')
plt.xlabel('Nilai Penjualan (Rp)')

plt.ticklabel_format(style='plain', axis='x')
plt.tight_layout()
plt.show()
```

```python
Q1 = df['Value'].quantile(0.25)
Q3 = df['Value'].quantile(0.75)

IQR = Q3 - Q1

batas_bawah = Q1 - (1.5 * IQR)
batas_atas = Q3 + (1.5 * IQR)

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Batas Bawah:", batas_bawah)
print("Batas Atas:", batas_atas)

outlier = df[
    (df['Value'] < batas_bawah) |
    (df['Value'] > batas_atas)
]

print("\nJumlah Outlier:", len(outlier))
print("\nContoh Data Outlier:")
display(outlier.head())
```

```python
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# 1. Bar Chart
sns.barplot(
    x=total_kategori.values,
    y=total_kategori.index,
    ax=axes[0, 0],
    color='skyblue'
)
axes[0, 0].set_title('Penjualan per Kategori')
axes[0, 0].set_xlabel('Total Penjualan')
axes[0, 0].set_ylabel('Kategori')

# 2. Line Chart
axes[0, 1].plot(
    penjualan_bulanan.index.astype(str),
    penjualan_bulanan.values,
    marker='o'
)
axes[0, 1].set_title('Tren Penjualan Bulanan')
axes[0, 1].set_xlabel('Bulan')
axes[0, 1].set_ylabel('Total Penjualan')
axes[0, 1].tick_params(axis='x', rotation=45)

# 3. Histogram
axes[1, 0].hist(df['Price'], bins=30, edgecolor='black')
axes[1, 0].set_title('Distribusi Harga Produk')
axes[1, 0].set_xlabel('Harga')
axes[1, 0].set_ylabel('Frekuensi')

# 4. Boxplot
sns.boxplot(x=df['Value'], ax=axes[1, 1], color='skyblue')
axes[1, 1].set_title('Sebaran Nilai Penjualan')
axes[1, 1].set_xlabel('Nilai Penjualan')

plt.tight_layout()
plt.show()
```
